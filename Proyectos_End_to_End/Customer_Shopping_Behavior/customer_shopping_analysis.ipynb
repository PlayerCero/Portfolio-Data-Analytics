```python
df["Review Rating"] = df.groupby("Category")["Review Rating"].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df = df.rename(columns={"purchase_amount_(usd)": "purchase_amount"})
```

```python
df.columns
```

```python
#create a column age_group
labels = ["Young Adult", "Adult", "Middle-aged", "Senior"]
df["age_group"] = pd.qcut(df["age"], q=4, labels = labels)
```

```python
df[["age","age_group"]].head(10)
```

```python
# create column purchase_frequency_days

frequency_mapping = {
    "Fortnightly": 14,
    "Weekly": 7,
    "Monthly": 30,
    "Quarterly": 90,
    "Bi-Weekly": 14,
    "Annually": 365,
    "Every 3 Months": 90
}

df["purchase_frequency_days"] = df["frequency_of_purchases"].map(frequency_mapping)
```

```python
df[["purchase_frequency_days","frequency_of_purchases"]].head(10)
```

```python
# Conexion a SQL Server (Windows Authentication)
from sqlalchemy import create_engine, text

SERVER = "DESKTOP-NMBIT88"
DATABASE = "shopping_analysis"

# Conexion inicial al motor (nivel servidor) para poder crear la base si no existe
master_conn_str = (
    f"mssql+pyodbc://@{SERVER}/master"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&trusted_connection=yes"
    "&TrustServerCertificate=yes"
)
master_engine = create_engine(master_conn_str, isolation_level="AUTOCOMMIT")

with master_engine.connect() as conn:
    conn.execute(text(
        f"IF NOT EXISTS (SELECT * FROM sys.databases WHERE name = '{DATABASE}') "
        f"CREATE DATABASE [{DATABASE}]"
    ))

print(f"Base de datos '{DATABASE}' lista.")
```

```python
# Engine apuntando ya a la base shopping_analysis
conn_str = (
    f"mssql+pyodbc://@{SERVER}/{DATABASE}"
    "?driver=ODBC+Driver+18+for+SQL+Server"
    "&trusted_connection=yes"
    "&TrustServerCertificate=yes"
)
engine = create_engine(conn_str)

with engine.connect() as conn:
    print(conn.execute(text("SELECT DB_NAME()")).scalar())
```

```python
# Subir el DataFrame limpio a SQL Server
df.to_sql("customer_shopping", con=engine, if_exists="replace", index=False)
print("Tabla customer_shopping cargada:", len(df), "filas")
```

```python
# Verificar desde SQL
pd.read_sql("SELECT TOP 10 * FROM customer_shopping", con=engine)
```

```python
# Ejemplo de analitica en SQL: gasto promedio y calificacion por categoria
query = """
SELECT
    category,
    COUNT(*) AS total_compras,
    AVG(purchase_amount) AS gasto_promedio,
    AVG(review_rating) AS rating_promedio
FROM customer_shopping
GROUP BY category
ORDER BY gasto_promedio DESC
"""
pd.read_sql(query, con=engine)
```
